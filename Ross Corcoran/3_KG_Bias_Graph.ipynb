The bias layer was represented using a NetworkX MultiDiGraph. This is a
allows more than one relationship between the same pairof nodes.

The final dataset contained 221 unique Head–Relation–Tail facts. 14
facts shared an existing Head and Tail but used a different relation label.
Using a MultiDiGraph preserved these separate extracted relationships
instead of overwriting one of them.

Gender group pilot test

```python
gender_groups = ["mná", "fir", "cailíní", "buachaillí"]


def find_gender_group(sentence):

    sentence = sentence.lower()
    for group in gender_groups:
      if group in sentence:
            return group

    #Return nothing if no gender group is found
    return None
```

```python
test_sentence = "Tá mná go dona ag matamaitic."

matched_group = find_gender_group(test_sentence)

print("Matched gender group:", matched_group)
```
